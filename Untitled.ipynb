```python
from nltk.util import ngrams

def splitting_dataframe(data):
    tokens=data.split()
    return tokens

data=clean_text_train.apply(splitting_dataframe)

def ngram_list(data,ngram_range):
    ngram=ngrams(data,ngram_range)
    ngram_list1=[]
    for ngram1 in ngram:
        ngram_list1.append(' '.join(ngram1))
    return ngram_list1

```
