```python
#unidecoding the  special characters in tweets

import contractions
def expand_text(text):
    return contractions.fix(text)

data["tweet"]= data["tweet"].apply(expand_text)
data.head()
```

```python
#HANDLING ACCENT

from unidecode import unidecode
def handling_accent(data):
    fixed_text=unidecode(data)

    return fixed_text

data["tweet"]= data["tweet"].apply(handling_accent)
data.head()
```

```python

nltk.download("stopwords",quiet=True)
```

```python
stopwordlist=stopwords.words('english')
stopwordlist.remove('no')
stopwordlist.remove('not')
stopwordlist.remove('nor')
```

```python
stopwordlist1 = ['a', 'about', 'above', 'after', 'again', 'ain', 'all', 'am', 'an',
             'and','any','are', 'as', 'at', 'be', 'because', 'been', 'before',
             'being', 'below', 'between','both', 'by', 'can', 'd', 'did', 'do',
             'does', 'doing', 'down', 'during', 'each','few', 'for', 'from',
             'further', 'had', 'has', 'have', 'having', 'he', 'her', 'here',
             'hers', 'herself', 'him', 'himself', 'his', 'how', 'i', 'if', 'in',
             'into','is', 'it', 'its', 'itself', 'just', 'll', 'm', 'ma',
             'me', 'more', 'most','my', 'myself', 'now', 'o', 'of', 'on', 'once',
             'only', 'or', 'other', 'our', 'ours','ourselves', 'out', 'own', 're','s',
                'same', 'she', "shes", 'should', "shouldve",'so', 'some', 'such',
             't', 'than', 'that', "thatll", 'the', 'their', 'theirs', 'them',
             'themselves', 'then', 'there', 'these', 'they', 'this', 'those',
             'through', 'to', 'too','under', 'until', 'up', 've', 'very', 'was',
             'we', 'were', 'what', 'when', 'where','which','while', 'who', 'whom',
             'why', 'will', 'with', 'won', 'y', 'you', "youd","youll", "youre",
             "youve", 'your', 'yours', 'yourself', 'yourselves']
```

```python
#removong stopwords

stopword=set(stopwordlist1)
def remove_stop_word(text):
    return " ".join([word for word in str(text).split() if word not in stopword])
data["tweet"]=data["tweet"].apply(lambda text :remove_stop_word(text))
data["tweet"].head()
```

```python
#removing punctuation

import string
punctuations=string.punctuation
def cleaning_text(text):
    translator = str.maketrans('', '', punctuations)
    return text.translate(translator)

data["tweet"]=data["tweet"].apply(lambda text: cleaning_text(text))
data["tweet"].head()
```

```python
#removing repeating characters

import re
def cleaning_repeating_character(text):
    return re.sub(r'(.)\1{2,}', r'\1\1', text)
data['tweet'] = data['tweet'].apply(lambda x: cleaning_repeating_character(x))
data['tweet'].head()
```

```python
#removing url

def removing_url(text):
    return re.sub('((www.[^s]+)|(https?://[^s]+))',' ',text)

data["tweet"]=data["tweet"].apply(lambda x: removing_url(x))
data["tweet"].head()
                                  
```

```python
#removing numbers

def removing_numbers(text):
    return re.sub('[0-9]+',' ',text)
data["tweet"]=data["tweet"].apply(lambda x: removing_numbers(x))

data["tweet"].head()
```

```python
#tokenizing the tweet

tokenizer = RegexpTokenizer(r'\w+')
data['tweet'] = data['tweet'].apply(tokenizer.tokenize)

data['tweet'].head()
```

```python
# perfoming stemmation on tweet

st = nltk.PorterStemmer()
def stemming(text):
    text = [st.stem(word) for word in text]
    return text
data["tweet"]= data["tweet"].apply(lambda x: stemming(x))
data["tweet"].head()
```

```python
#performing lemmitization

lm =nltk.WordNetLemmatizer()
def lemmatizer_on_text(text):
    text = [lm.lemmatize(word) for word in text]
    return text
data['tweet'] = data['tweet'].apply(lambda x: lemmatizer_on_text(x))
data['tweet'].head()
```

```python
x=data.tweet
y=data.target
```

```python
# training and testing

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.05,random_state=26105111)
```

```python
x_train.shape,y_train.shape,x_test.shape
x_train=x_train.astype(str)
x_test=x_test.astype(str)

x_test.head()
```

```python
#tfid vectorisation

vectoriser=TfidfVectorizer(ngram_range=(1,2), max_features=500000)
vectoriser.fit(x_train)
print(' No. of feature_words: ', len(vectoriser.get_feature_names_out()))
```

```python
x_train=vectoriser.transform(x_train)
x_test=vectoriser.transform(x_test)
```

```python
#CONFUSION MATRIX AND CLASSIFICATION REPORT

import seaborn as sns
def model_Evaluate(model):
    y_pred = model.predict(x_test)
    print(classification_report(y_test, y_pred))
    
    cf_matrix = confusion_matrix(y_test, y_pred)
    categories = ['Negative','Positive']
    group_names = ['True Negative','False Positive', 'False Negative','True Positive']
    group_percentages = ['{0:.2%}'.format(value) for value in cf_matrix.flatten() / np.sum(cf_matrix)]
    labels = [f'{v1}n{v2}' for v1, v2 in zip(group_names,group_percentages)]
    labels = np.asarray(labels).reshape(2,2)
    sns.heatmap(cf_matrix, annot = labels, cmap = 'Blues',fmt = '',
    xticklabels = categories, yticklabels = categories)

   
    plt.xlabel("Predicted values", fontdict = {'size':14}, labelpad = 10)
    plt.ylabel("Actual values" , fontdict = {'size':14}, labelpad = 10)
    plt.title ("Confusion Matrix", fontdict = {'size':18}, pad = 20)
    plt.show()
```

```python

#MODEL BUILDING
```

```python
# FITTING IN BERNOULIMNB MODEL

BNBmodel = BernoulliNB()
BNBmodel.fit(x_train, y_train)
model_Evaluate(BNBmodel)
y_pred1 = BNBmodel.predict(x_test)
```

```python
fpr,tpr,threshold=roc_curve(y_test,y_pred1)

roc_auc=auc(fpr,tpr)

plt.figure(figsize=(8,6))
plt.plot(fpr,tpr,color='darkorange',lw=2,
         label=f"ROc curve( Auc = {roc_auc:.2f})")
plt.plot([0,1],[0,2],color='navy',lw=2,linestyle='--',label='Random')
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('Roc curve for berounli MNB model '.format(accuracy=200))
plt.legend(loc="lower right")
plt.show()
```

```python
# FITTING IN LINEAR SVC MODEL


svcmodel=LinearSVC()
svcmodel.fit(x_train,y_train)
model_Evaluate(svcmodel)
y_pred2=svcmodel.predict(x_test)
```

```python
tpr,fpr,threshold=roc_curve(y_test,y_pred2)
roc_auc = auc(tpr,fpr)
plt.plot(tpr,fpr,color="darkorange",lw=1,
          label=f"ROc curve( Auc = {roc_auc:.2f})")
plt.plot([0,1],[0,2],color='navy',lw=2,linestyle='--',label='Random')
plt.xlabel("false postve rate")
plt.ylabel("true postive rate")
plt.title( "ROC CURVE linear SVC model")
plt.legend(loc="lower right")
plt.show()








```

```python
# FITTING IN LOGISTIC REGRESSION MODEL


Lgmodel=LogisticRegression(max_iter=1000,n_jobs=-1)

Lgmodel.fit(x_train,y_train)
model_Evaluate(Lgmodel)
y_pred3=Lgmodel.predict(x_test)
```

```python
tpr,fpr,threshold=roc_curve(y_test,y_pred3)
roc_auc = auc(tpr,fpr)
plt.plot(tpr,fpr,color="darkorange",lw=1,
          label=f"ROC curve( Auc = {roc_auc:.2f})")
plt.plot([0,1],[0,2],color='navy',lw=2,linestyle='--',label='Random')
plt.xlabel("false postve rate")
plt.ylabel("true postive rate")
plt.title( "ROC CURVE logistic regression model ")
plt.legend(loc="lower right")
plt.show()
```
